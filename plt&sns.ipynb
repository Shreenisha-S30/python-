```python
#matplot pyplot
#we can use some special symobols called marker to plot graphs like o, *, . , , , x, X etc
import matplotlib.pyplot as plt
import numpy as np

x = np.array([0, 4, 6, 8])
y = np.array([0, 50, 100, 250])

plt.plot(x,y, 'o')
plt.title("matplot demo")
plt.show()
```

```python
# can use the keyword argument marker to emphasize each point with a specified marker = ('o' Circle,	'*'	Star,	'.'	Point,	','	Pixel,	'x'	X,	'X'	X (filled),	'+'	Plus,	'P'	Plus (filled),	's'	Square,	'D'	Diamond,	'd'	Diamond (thin),	'p'	Pentagon,	'H'	Hexagon,	'h'	Hexagon,	'v'	Triangle Down,	'^'	Triangle Up,	'<'	Triangle Left,	'>'	Triangle Right,	'1'	Tri Down,	'2'	Tri Up,	'3'	Tri Left,	'4'	Tri Right,	'|'	Vline,	'_'	Hline )
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([5,14,3,19,6])
xpoints = np.array([16, 7, 3, 19, 3])

plt.plot(xpoints, ypoints, marker = 'o')

plt.show()
```

```python
#this is also called as format string marker|line|color
#can use the keyword argument markersize or the shorter version, ms to set the size of the markers
#You can use the keyword argument markeredgecolor or the shorter mec to set the color of the edge of the markers
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, 'o-.m', ms = 20, mec = 'k')
plt.show()
```

```python
#You can use the keyword argument markerfacecolor or the shorter mfc to set the color inside the edge of the marker

import matplotlib.pyplot as plt
import numpy as np

yaxis = np.array([5,3,10,1])

plt.plot(yaxis, '*:c', ms = 25, mfc = 'r')

plt.show()
```

```python
#using mec and mfc arguments to color the entire marker

import matplotlib.pyplot as plt
import numpy as np

xaxis = np.array([6,2,8,5])

plt.plot(xaxis, marker = 'o', ms = 15, mec = 'hotpink', mfc = 'hotpink')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

xaxis = np.array([4, 2, 17, 9])

plt.plot(xaxis, ls = 'dotted', c = 'r', linewidth = '10')      #linestyle can be written as ls and color as c

plt.show()
```

```python
#scatter
import matplotlib.pyplot as plt
import numpy as np

x = np.array([5,7,8,7,14,17,2,9,4,11,12,9,6])
y = np.array([99,96,107,88,101,96,103,67,94,78,97,85,106])
colors = np.array(["red","green","blue","yellow","pink","black","orange","purple","beige","brown","gray","cyan","magenta"])
sizes = np.array([20,50,100,200,500,1000,60,90,100,300,600,800,75])

plt.scatter(x, y, c = colors, s = sizes, alpha = 0.6)
plt.colorbar()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.random.randint(100, size=(100))
y = np.random.randint(100, size=(100))
colors = np.random.randint(100, size=(100))
sizes = 10 * np.random.randint(100, size=(100))

plt.scatter(x, y, c=colors, s=sizes, alpha=0.6, cmap='nipy_spectral')

plt.colorbar()

plt.show()
```
